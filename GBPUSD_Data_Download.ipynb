```python
import yfinance as yf
import pandas as pd

# Download GBP/USD hourly data
# yfinance returns max 60 days per call — loop to get more
ticker = yf.Ticker("GBPUSD=X")

# Get 60-day window (hourly)
df = ticker.history(period="600d", interval="1h")
df = df[['Open', 'High', 'Low', 'Close', 'Volume']]
df.index = pd.to_datetime(df.index)

# Save to CSV for reuse
df.to_csv("GBPUSD_hourly_600Days_data.csv")
print(df.head(), df.shape)
```
